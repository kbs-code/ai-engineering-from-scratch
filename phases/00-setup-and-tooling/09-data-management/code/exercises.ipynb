Ex 02

```python
import time

dataset = load_dataset("allenai/c4", "en", split="train", streaming=True)
start = time.time()
count = 0
for example in dataset:
    count += 1
    if time.time() - start >= 10:
        break
print(f"Processed {count} examples in 10 seconds.")
```

Ex 03

```python
import os

dataset = load_dataset("google/WikiProfile", split="train")

dataset.to_csv("c4.csv")
dataset.to_parquet("c4.parquet")

csv_size = os.path.getsize("c4.csv")
parquet_size = os.path.getsize("c4.parquet")

print(f"CSV:     {csv_size / 1024:.1f} KB")
print(f"Parquet: {parquet_size / 1024:.1f} KB")
print(f"Parquet is {csv_size / parquet_size:.1f}x smaller")
```

```python
dataset = load_dataset("google/WikiProfile", split="train")
split_1 = dataset.train_test_split(test_size=0.3, seed=42)
split_2 = split_1["test"].train_test_split(test_size=0.5, seed=42)
final = {
    "train": split_1["train"],      
    "validation": split_2["train"], 
    "test": split_2["test"],        
}

print(len(final["train"]))      
print(len(final["validation"])) 
print(len(final["test"]))     
```
